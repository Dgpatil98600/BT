```python
import random

# 1. Helper function to find Greatest Common Divisor
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# 2. Helper function to find Modular Inverse (d)
# This uses the Extended Euclidean Algorithm
def mod_inverse(e, phi):
    def extended_gcd(a, b):
        if a == 0:
            return b, 0, 1
        else:
            gcd, x, y = extended_gcd(b % a, a)
            return gcd, y - (b // a) * x, x
            
    g, x, y = extended_gcd(e, phi)
    
    if g != 1:
        raise Exception('Modular inverse does not exist')
    else:
        return x % phi

# 3. Key Generation
def generate_keypair(p, q):
    n = p * q
    phi = (p - 1) * (q - 1)

    # Choose e
    e = random.randrange(1, phi)
    g = gcd(e, phi)
    while g != 1:
        e = random.randrange(1, phi)
        g = gcd(e, phi)

    # Calculate d
    d = mod_inverse(e, phi)
    
    # Return ((Public Key), (Private Key))
    return ((e, n), (d, n))

def encrypt(pk, plaintext):
    # Unpack the key into key and n
    key, n = pk
    # Convert each letter in the plaintext to numbers based on the character using a^b mod m
    cipher = [pow(ord(char), key, n) for char in plaintext]
    return cipher

def decrypt(pk, ciphertext):
    # Unpack the key into key and n
    key, n = pk
    # Generate the plaintext based on the ciphertext and key using a^b mod m
    plain = [chr(pow(char, key, n)) for char in ciphertext]
    return ''.join(plain)

# --- EXECUTION ---

# We use small primes here for demonstration speed. 
# In reality, these primes would be hundreds of digits long.
p = 61
q = 53

public_key, private_key = generate_keypair(p, q)

print(f"Public Key: {public_key}")
print(f"Private Key: {private_key}")

message = "HELLO"
print(f"\nOriginal Message: {message}")

# Encrypt
encrypted_msg = encrypt(public_key, message)
print(f"Encrypted (list of integers): {encrypted_msg}")

# Decrypt
decrypted_msg = decrypt(private_key, encrypted_msg)
print(f"Decrypted Message: {decrypted_msg}")
```
